# General

### Import

````python
from matplotlib import pyplot as plt
import cv2 as cv
from mtcnn import MTCNN
````

### Variable

````python
root_dir = '/Volumes/data/documents/ai_document'
````

# Detect

### Instance

````python
detector = MTCNN()
````

### Read



````python
address =f"{root_dir}/pic/maryam.jpg"
img = cv.imread(address)
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
````

### Display 

````python
plt.imshow(img)
plt.show()
````

### Detect

````python
img_detect = detector.detect_faces(img)[0]
x, y, w, h = img_detect["box"]
confidence = img_detect["confidence"]
keypoints = img_detect["keypoints"]
````

### Add items

````python
img_2 = cv.imread(address)
img_2 = cv.cvtColor(img_2, cv.COLOR_BGR2RGB)

cv.rectangle(img_2, (x, y), (x+w, y+h), (255,0,0), 3)

confidence = "prob: {:.2f}".format(confidence*100)
cv.putText(img_2, confidence, (x, y-20), cv.FONT_HERSHEY_PLAIN, 1, (0,255,0), 2)

for key, value in keypoints.items():
    cv.circle(img_2, value, 5, (255,0,0), -1)
````

### Display

````python
plt.imshow(img_2)
plt.show()
````

# Split

````python
face = img[y:y+h, x:x+w]
````

````python
plt.imshow(face)
plt.show()
````

# Save

````python
cv.imwrite(f"{root_dir}/pic/maryam_face.jpg", cv.cvtColor(face, cv.COLOR_RGB2BGR))
````
