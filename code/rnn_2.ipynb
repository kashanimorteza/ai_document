# General

Import

````python
import os
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
import keras as ks
import tensorflow as tf
````

Variable

````python
MAX_FEATURES = 20000
MAXLEN = 80
BATCH_SIZE = 100
EPOCHS = 15
````

# Dataset

Read

````python
(x_train, y_train), (x_test, y_test) = ks.datasets.imdb.load_data(num_words=MAX_FEATURES)
````

Normalize

````python
x_train = ks.utils.pad_sequences(x_train, maxlen=MAXLEN)
x_test = ks.utils.pad_sequences(x_test, maxlen=MAXLEN)
````

Shape

````python
print("x_train : {}".format(x_train.shape))
print("x_test  : {}".format(x_test.shape))
print("y_train : {}".format(y_train.shape))
print("y_test  : {}".format(y_test.shape))
````

Label

````python
print("Label  : {}".format(y_train[0]))
````

Data

````python
print("Data  : {}".format(x_train[0]))
````

# Model

````python
mdl = ks.Sequential()
mdl.add(ks.layers.Embedding(MAX_FEATURES, 128))
mdl.add(ks.layers.SimpleRNN(128))
mdl.add(ks.layers.Dense(1, activation='sigmoid'))
mdl.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
````

# Train

````python
h = mdl.fit(
    x_train,
    y_train,
    batch_size=BATCH_SIZE,
    epochs = EPOCHS,
    validation_data=(x_test, y_test)
)
````

# Evaluation

````python
score, accuracy = mdl.evaluate(
    x_test,
    y_test,
    batch_size=BATCH_SIZE
)
print("Score: {:.2f} | Accuracy: {:.2f}".format(score, accuracy))
````
