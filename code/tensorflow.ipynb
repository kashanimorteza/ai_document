## Decorators

### @tf.function

````python
def eager_function1():
    a = tf.constant([1])
    b = tf.constant([2])
    print(a + b) 
    
eager_function1()
````

````python
@tf.function
def eager_function2():
    a = tf.constant([1])
    b = tf.constant([2])
    print(a + b)

eager_function2()
````

````python
graph_function = tf.function(eager_function1)
graph_function()
````
