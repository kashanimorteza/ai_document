# Class

````python
class GNet(ks.models.Model):
    def __init__(self):
        super().__init__()
        self.conv1 = ks.layers.Conv2D(32, (3, 3), activation='relu')
        self.conv2 = ks.layers.Conv2D(64, (3, 3), activation='relu')
        self.pool1 = ks.layers.MaxPooling2D((2, 2))
        self.drop1 = ks.layers.Dropout(0.25)
        self.flat = ks.layers.Flatten()
        self.dense1 = ks.layers.Dense(128, activation='relu')
        self.drop2 = ks.layers.Dropout(0.5)
        self.dense2 = ks.layers.Dense(10, activation="softmax")
````

# Call

````python
def call(self, inputs):
    x = self.conv1(inputs)
    x = self.conv2(x)
    x = self.pool1(x)
    x = self.drop1(x)
    x = self.flat(x)
    x = self.dense1(x)
    x = self.drop2(x)
    x = self.dense2(x)

    return x
````
