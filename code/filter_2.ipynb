# Action

### Input

````python
x = np.array(
    [
        [3, 5, 2, 7],
        [4, 1, 3, 8],
        [6, 3, 8, 2],
        [9, 6, 1, 5]
    ]
)

x = x.reshape(1, 4, 4, 1)
````

### weights

````python
weights = [
    np.array(
        [
            [
                [
                    [1]
                ],
                [
                    [2]
                ],
                [
                    [1]
                ]
            ],
            [
                [
                    [2]
                ],
                [
                    [1]
                ],
                [
                    [2]
                ]
            ],
            [
                [
                    [1]
                ],
                [
                    [1]
                ],
                [
                    [2]
                ]
            ]
        ]
    ),
    np.asarray([0])
]
````

### Decrease : Conv2D

````python
mdl = models.Sequential()
mdl.add(layers.Input(shape=(4, 4, 1)))
mdl.add(layers.Conv2D(1, (3, 3), strides=(1,1), padding="valid"))
mdl.set_weights(weights)
out_1 = mdl.predict(x)
print(x.reshape(4,4))
print("\n")
print(out_1.reshape(2,2))
````

### Increase : Conv2DTranspose

````python
mdl = models.Sequential()
mdl.add(layers.Input(shape=(2, 2, 1)))
mdl.add(layers.Conv2DTranspose(1, (3, 3), strides=(1,1), padding="valid"))
mdl.set_weights(weights)
out_2 = mdl.predict(out_1)
print(x.reshape(4,4))
print("\n")
print(out_2.reshape(4 ,4))
````
