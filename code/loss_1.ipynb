# General

````python
import numpy as np
import tensorflow as tf
import keras as ks
````

# General

````python
y_true = np.array([8, 8, 8, 8])
y_pred = np.array([6, 6, 6, 6])
print("y_true: {} | shape:{}".format(y_true, y_true.shape))
print("y_pred: {} | shape:{}".format(y_pred, y_pred.shape))

a = y_true - y_pred
print("a:      {} | shape:{}".format(a, a.shape))

b = tf.square(a)
print("b:      {} | shape:{}".format(b, b.shape))

# -1 = Last dimension of the array
c = tf.reduce_mean(b, axis=-1)
print("c:      {} | shape:{}".format(c, c.shape))
````

# MSE

data

````python
y_true = np.array([8, 8, 8, 8], dtype=np.float32)
y_pred = np.array([6, 6, 6, 6], dtype=np.float32)
````

Way-1

````python
error = tf.square(y_true - y_pred)
loss = tf.reduce_mean(error, axis=-1)
print("loss: {}".format(loss))
````

Way-2

````python
loss = ks.losses.mean_squared_error(y_true, y_pred)
print("loss: {}".format(loss))
````

Way-3

````python
class myMSE(ks.losses.Loss):
    def __init__(self):
        super().__init__()

    def call(self, y_true, y_pred):
        error = tf.square(y_true - y_pred)
        loss = tf.reduce_mean(error, axis=-1)
        return loss

fn = myMSE()
loss = fn(y_true, y_pred)
print("loss: {}".format(loss))
````

# MAE

Data

````python
y_true = np.array([8, 8, 8, 8], dtype=np.float32)
y_pred = np.array([6, 6, 6, 6], dtype=np.float32)
````

Way-1

````python
error = np.abs(y_true - y_pred)
loss = np.mean(error)
print("loss: {}".format(loss))
````

Way-2

````python
loss = ks.losses.mean_absolute_error(y_true, y_pred)
print("loss: {}".format(loss))
````

Way-3

````python
class myMAE(ks.losses.Loss):
    def __init__(self):
        super().__init__()

    def call(self, y_true, y_pred):
        error = np.abs(y_true - y_pred)
        loss = np.mean(error)
        return loss

fn = myMAE()
loss = fn(y_true, y_pred)
print("loss: {}".format(loss))
````

# Huber

Data

````python
y_true = np.array([8, 8, 8, 8], dtype=np.float32)
y_pred = np.array([6, 6, 6, 6], dtype=np.float32)
````

Way-1

````python
hyper_parameter = 1.0
error = y_true - y_pred
is_small_error = tf.abs(error) <= hyper_parameter
small_error_loss = tf.square(error)/2.0
big_error_loss = hyper_parameter * (tf.abs(error) - (0.5 * hyper_parameter))
loss = tf.where(is_small_error, small_error_loss, big_error_loss )
loss = tf.reduce_mean(loss)
print("loss: {}".format(loss))
````

Way-2

````python
huber = ks.losses.Huber(delta=1.0)
loss = huber(y_true, y_pred)
print("loss: {}".format(loss))
````

Way-3

````python
class myHuber(ks.losses.Loss):
    def __init__(self, hyper_parameter):
        super().__init__()
        self.hyper_parameter = hyper_parameter

    def call(self, y_true, y_pred):
        error = y_true - y_pred
        is_small_error = tf.abs(error) <= self.hyper_parameter
        small_error_loss = tf.square(error)/2.0
        big_error_loss = self.hyper_parameter * (tf.abs(error) - (0.5 * self.hyper_parameter))
        loss = tf.where(is_small_error, small_error_loss, big_error_loss )
        loss = tf.reduce_mean(loss)
        return loss

fn = myHuber(hyper_parameter=1.0)
loss = fn(y_true, y_pred)
print("loss: {}".format(loss))
````
