## Import

````python
import glob
import numpy as np
import cv2 as cv
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score
from joblib import dump, load
````

## Variable

````python
root_dir = '/Volumes/data/documents/ai_document'
````

## Read

````python
img_list = []
for address in glob.glob(f"{root_dir}/file/dataset/gender/**/*.jpg"):
    img = cv.imread(address)
    img_list.append(img)
````

## Label

````python
labels = []
for address in glob.glob(f"{root_dir}/file/dataset/gender/**/*.jpg"):
    label = address.split('/')[-2].split('.')[0]
    labels.append(label)
````

## Preprocessing

````python
img_resize = []
for img in img_list:
    img = cv.resize(img, (32, 32))
    img_resize.append(img)
````

## Normalization

````python
img_normalize = []
for img in img_resize:
    img = img/255.0
    img_normalize.append(img)
````

## Flat

````python
img_flat = []
for img in img_normalize:
    img = img.flatten()
    img_flat.append(img)
````

## Convert

````python
data_list = np.array(img_flat)
````

## Separate tarin and test

````python
x_train, x_test, y_train, y_test = train_test_split(data_list, labels, test_size=0.2)
````

## Display

````python
print(data_list.shape)
print(x_train.shape)
print(x_test.shape)
````

## Train

````python
clf = SGDClassifier()
clf.fit(x_train, y_train)
y_pred = clf.predict(x_test)
````

## Save model

````python
dump(clf, "/Volumes/data/documents/ai_document/file/dataset/gender/model.z")
````

## Accuracy

````python
accuracy = clf.score(x_test, y_test)
print("Accuracy: {:.2f}".format(accuracy * 100))
````
