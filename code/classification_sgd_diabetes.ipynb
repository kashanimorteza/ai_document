## Import

````python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, zero_one_loss
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier
````

## Import dataset

````python
dataset = pd.read_csv("diabetes.csv")
````

## Display dataset

````python
print(dataset.head(5))
rows, cols = dataset.shape
print("rows: {}, cols:{}".format(rows, cols))
````

## Missing values

````python
zero_not_accepted = ['Glucose', 'BloodPressure','SkinThickness', 'BMI', 'Insulin']
for coloumn in zero_not_accepted:
  dataset[coloumn] = dataset[coloumn].replace(0, np.nan)
  mean = int(dataset[coloumn].mean(skipna=True))
  dataset[coloumn]=dataset[coloumn].replace(np.nan, mean)
````

## Separate data and label

````python
x = dataset.iloc[:, :8]
y = dataset.iloc[:, 8]
print(x.head(5))
print(y.head(5))
````

## Separate tarin and test

````python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
````

````python
print("x_train: {}".format(x_train.shape))
print("x_test: {}".format(x_test.shape))
print("y_train: {}".format(y_train.shape))
print("y_test: {}".format(y_test.shape))
````

## Normalization

````python
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train)
x_test = sc_x.transform(x_test)
````

## Train

````python
#---using of Logistic
mdl = SGDClassifier(loss="log_loss")
mdl.fit(x_train, y_train)
````

## Predict

````python
y_pred = mdl.predict(x_test)
df_pred = pd.DataFrame({"Actual": y_test, "Predicted": y_pred})
df_pred["Correct"] = df_pred["Actual"] == df_pred["Predicted"]
print(df_pred)
````

## Accuracy and Loss

````python
accuracy = accuracy_score(y_test, y_pred) * 100
loss = zero_one_loss(y_test, y_pred) * 100
accuracy2 = mdl.score(x_test, y_test) * 100
print("Accuracy: {:.2f} | Loss: {:.2f} | accuracy2: {:.2f}".format(accuracy, loss, accuracy2))
````

## Weight and Bias

````python
weightes = mdl.coef_
bias = mdl.intercept_
epochs = mdl.n_iter_
print("epochs: {} | Bias: {} | Weightes: {}".format(epochs, bias, weightes))
````

## Prob

````python
prob = mdl.predict_proba(x_test)
print(prob)
````
