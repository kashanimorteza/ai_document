## Import

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, zero_one_loss
````

## Import dataset

````python
dataset = pd.read_csv("iris.csv", header=None, names = ['sepal length', 'sepal width', 'petal length', 'petal width', 'label'])
````

## Display dataset

````python
print(dataset.head(5), '\n')
rows, cols = dataset.shape
print("rows: {}, cols:{}".format(rows, cols))
````

## Separate data and label

````python
data = dataset.iloc[:, :4]
label = dataset.iloc[:, 4]
print(data.head(5), '\n')
print(label.head(5))
````

## Separate tarin and test

````python
x_train, x_test, y_train, y_test = train_test_split(data, label, test_size=0.2)
````

````python
print("x_train: {}".format(x_train.shape))
print("x_test: {}".format(x_test.shape))
print("y_train: {}".format(y_train.shape))
print("y_test: {}".format(y_test.shape))
````

## Train

````python
k=5
mdl = KNeighborsClassifier(n_neighbors=k)
mdl.fit(x_train, y_train)
````

## Predict

````python
y_pred = mdl.predict(x_test)
df_pred = pd.DataFrame({"Actual": y_test, "Predicted": y_pred})
df_pred["Correct"] = df_pred["Actual"] == df_pred["Predicted"]
print(df_pred)
````

## Accuracy and Loss

````python

accuracy = accuracy_score(y_test, y_pred) * 100
loss = zero_one_loss(y_test, y_pred) * 100
print("Accuracy: {:.2f} | Loss: {:.2f}".format(accuracy, loss))
````
