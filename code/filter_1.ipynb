# General

### Import

````python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
````

### Variable

````python
root_dir = '/Volumes/data/documents/ai_document'
````

# Action

### Image

````python
img = cv.imread(f"{root_dir}/pic/lena.png")
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.imshow(img)
plt.show()
````

### x_direction_kernel

````python
x_direction_kernel = np.array(
    [
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ]
)

img_x = cv.filter2D(img, -1, x_direction_kernel)

plt.imshow(img_x)
plt.show()
````

### y_direction_kernel

````python
y_direction_kernel = np.array(
    [
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ]
)

img_y = cv.filter2D(img, -1, y_direction_kernel)

plt.imshow(img_y)
plt.show()
````
