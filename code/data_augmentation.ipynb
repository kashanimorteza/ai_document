# General

### Import

````python
import glob, subprocess
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator
````

### Variable

````python
root_dir = '/Volumes/data/documents/ai_document'
````

# Action

### Read image

````python
img = cv.imread(f"{root_dir}/pic/001.jpg")
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

print(img.shape)
plt.imshow(img)
plt.show()
````

### Data Augmentation

````python
aug = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)
````

````python
img = np.array([img])
image_gen = aug.flow(
    img,
    batch_size=1,
    save_to_dir=f"{root_dir}/pic/data_ugmentation",
    save_format='jpeg',
    save_prefix='cv'
)
````

### Remove all image

````python
subprocess.run(f"find {root_dir}/pic/data_ugmentation -mindepth 1 -delete", shell=True)
````
