# General

### Import

````python
from matplotlib import pyplot as plt
import cv2 as cv
from mtcnn import MTCNN
````

### Variable

````python
root_dir = '/Volumes/data/documents/ai_document'
````

# Action

### Instance

````python
detector = MTCNN()
````

### Read



````python
imgs = []
address =f"{root_dir}/video/001.mp4"
cap = cv.VideoCapture(address)

while(cap.isOpened()):
    _, img = cap.read()
    if img is None : break
    img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    imgs.append(img)

print("Images: {}".format(len(imgs)))
````

### Display 

````python
plt.imshow(imgs[50])
plt.show()
````

### Detect

````python
img_detects = []
image_number = 0
for img in imgs:
    image_number +=1
    if image_number > 50:
        img_detect = detector.detect_faces(img)
        if len(img_detect)>0:
            #---Detect
            img_detect = img_detect[0]
            x, y, w, h = img_detect["box"]
            confidence = img_detect["confidence"]
            keypoints = img_detect["keypoints"]
            #---Add items
            cv.rectangle(img, (x, y), (x+w, y+h), (0,255,0), 1)
            confidence = "prob: {:.2f}".format(confidence*100)
            cv.putText(img, str(image_number), (x, y+40), cv.FONT_HERSHEY_PLAIN, 3, (0,255,0), 2)
            cv.putText(img, confidence, (x, y+90), cv.FONT_HERSHEY_PLAIN, 3, (0,255,0), 2)
            for key, value in keypoints.items() : cv.circle(img, value, 5, (0,255,0), -1)
            #---Display
            img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
            cv.imshow("face", img)
            if cv.waitKey(5) == ord("q") : break
            #---Add list
            img_detects.append(img)
            print("image_number: {}".format(image_number))

cv.destroyAllWindows()
print("img_detects: {}".format(len(img_detects)))
````
