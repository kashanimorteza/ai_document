## Import

````python
import cv2 as cv
import glob
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from joblib import dump, load
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
````

## Read



````python
img_list = []
for address in glob.glob('/Volumes/data/documents/ai_document/file/dataset/fire_dataset/**/*.png', recursive=True):
    img = cv.imread(address)
    img_list.append(img)
````

## Label

````python
lables = []
for address in glob.glob('/Volumes/data/documents/ai_document/file/dataset/fire_dataset/**/*.png', recursive=True):
    label = address.split('/')[-1].split('.')[0]
    lables.append(label)
````

## Preprocessing

````python
img_resize = []
for img in img_list:
    img = cv.resize(img, (32, 32))
    img_resize.append(img)
````

## Normalization

````python
img_normalize = []
for img in img_resize:
    img = img/255.0
    img_normalize.append(img)
````

## Flat

````python
img_flat = []
for img in img_normalize:
    img = img.flatten()
    img_flat.append(img)
````

## Convert

````python
data_list = np.array(img_flat)
````

## Separate tarin and test

````python
x_train, x_test, y_train, y_test = train_test_split(data_list, lables, test_size=0.2)
````

## Display

````python
print(data_list.shape)
print(x_train.shape)
print(x_test.shape)
````

## Train

````python
clf = KNeighborsClassifier()
clf.fit(x_train, y_train)
y_pred = clf.predict(x_test)
````

## Save model

````python
dump(clf, "fire_detector.z")
````

## Use model

````python
clf = load('/Volumes/data/documents/ai_document/file/dataset/fire_dataset/fire_detector.z')
for address in glob.glob('/Volumes/data/documents/ai_document/file/dataset/fire_dataset/predict/*.jpg'):
    img = cv.imread(address)
    img = cv.resize(img, (32, 32))
    img = img/255.0
    img = img.flatten()
    img = np.array([img])
    out = clf.predict(img)[0]
    name = address.split('/')[-1].split('.')[0]
    print("{} : {}".format(name, out))
````

## Accuracy

````python
accuracy = clf.score(x_test, y_test)
print("Accuracy: {:.2f}".format(accuracy * 100))
````

````python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: {:.2f}".format(accuracy * 100))

precision = precision_score(y_test, y_pred, pos_label="fire")
print("Precision: {:.2f}".format(accuracy * 100))

recall = recall_score(y_test, y_pred, pos_label="fire")
print("Recall: {:.2f}".format(accuracy * 100))

f_score = f1_score(y_test, y_pred, pos_label="fire")
print("Recall: {:.2f}".format(f_score * 100))

#print(confusion_matrix(y_test, y_pred))
#tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
````
