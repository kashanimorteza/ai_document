# General

````python
import numpy as np
import tensorflow as tf
import keras as ks
````

# Data

````python
x = np.array([-2.0, -1.5, -0.5, 0, 0.5, 1.5, 2], dtype=np.float32)
````

# Way-1

````python
activation = []
for i in x:
    if i > 0: 
        activation.append(i)
    else:
        activation.append(0)
        
print("activation:", np.array(activation, dtype=np.float32))
````

# Way-2

````python
activation = tf.maximum(0.0, x)

print("activation: {}".format(activation))
````

# Way-3

````python
activation = tf.nn.relu(x)

print("activation: {}".format(activation))
````

# Way-4

````python
def myRelu(x):
    return tf.maximum(0.0, x)

input_layer = ks.layers.Input(shape=(1,1))
x = ks.layers.Lambda(myRelu)(input_layer)

model = ks.models.Model(inputs=input_layer, outputs=x, name="SimpleReLU")

model.summary()
````

# Way-5

````python
input_layer = ks.layers.Input(shape=(1,1))
x = ks.layers.Lambda(lambda t: tf.maximum(0.0, t))(input_layer)

model = ks.models.Model(inputs=input_layer, outputs=x, name="SimpleReLU")

model.summary()
````
