# Dataset

Variable

````python
NUM_EXAMPLES = 1000
w = tf.Variable(2.0)
b = tf.Variable(1.0)
TRUE_w = 3.0
TRUE_b = 2.0
````

Read

````python
x = tf.random.normal(shape=(NUM_EXAMPLES,))
y = (x * TRUE_w) + TRUE_b
p = (x * w) + b
````

````python
print("x :", x[0])
print("y :", y[0])
print("p :", p[0])
````

Plot

````python
plt.scatter(x, y, c='b', marker='.', label = "Real Data")
plt.scatter(x, p, c='r', marker='.', label = "Predicted Data")
plt.legend()
plt.show()
````

# Model

Loss

````python
def loss(p, y):
	return tf.reduce_mean(tf.square(p - y))
````

# Train

Variable

````python
EPOCHS = 75
BATCH_SIZE = 5
LEARNING_RATE = 0.01
````

Method

````python
def train(y, learning_rate):
    with tf.GradientTape() as t:
        p = w * x + b
        current_loss = loss(p, y)
    dw, db = t.gradient(current_loss, [w, b])
    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)
    return current_loss
````

Action

````python
list_w, list_b, losses = [], [], []
epochs = range(EPOCHS)
for epoch in epochs:
    list_w.append(w.numpy())
    list_b.append(b.numpy())
    current_loss = train(y, LEARNING_RATE)
    losses.append(current_loss)
    print(f'Epoch {epoch}: w={list_w[-1]:.2f} b={list_b[-1]:.2f}, loss={current_loss:.5f}')
````

# Evaluation

w - True_w

````python
plt.plot(epochs, list_w, 'r')
plt.plot([TRUE_w] * len(epochs), 'r.')
plt.legend(['w', 'True w'])
plt.show()
````

b - True_b

````python
plt.plot(epochs, list_b, 'b')
plt.plot([TRUE_b] * len(epochs), 'b.')
plt.legend(['b', 'True b'])
plt.show()
````

Loss - weights

````python
plt.plot(list_w, losses, 'r')
plt.plot(list_w, losses, 'bo')
plt.xlabel('weights') 
plt.ylabel('Loss')
plt.show()
````

Loss - Bias

````python
plt.plot(list_b, losses, 'r')
plt.plot(list_b, losses, 'bo')
plt.xlabel('Bias')
plt.ylabel('Loss')
plt.show()
````
